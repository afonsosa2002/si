### Exercise 1: NumPy array Indexing/Slicing

### 1.1) Load the "iris.csv" file:

```python
from sklearn.datasets import load_iris
import pandas as pd

iris_data = load_iris()


iris_df = pd.DataFrame(
    data=iris_data.data, 
    columns=iris_data.feature_names  
)

iris_df['species'] = iris_data.target

iris_df['species'] = iris_df['species'].map(
    {0: 'setosa', 1: 'versicolor', 2: 'virginica'}
)

iris = iris_df
```

### 1.2) Select the penultimate independent variable and find its dimension:

```python
penultimate_var = iris.iloc[:, -2]
dimension = penultimate_var.shape
print(f"Dimension of the array: {dimension}")
```

### 1.3) Select the last 10 samples and calculate the mean for each independent variable:

```python
last_10_samples = iris.iloc[-10:, :-1] 

mean_values = last_10_samples.mean()
print("Mean of the last 10 samples for each independent variable:")
print(mean_values)
```

### 1.4) Select all samples with values ≤ 6 for all independent variables:


```python
filtered_samples = iris[(iris.iloc[:, :-1] <= 6).all(axis=1)]
num_samples = filtered_samples.shape[0]
print(f"Number of samples with all independent variables ≤ 6: {num_samples}")
```

### 1.5) Select samples with a class/label different from 'Iris-setosa':

```python
non_setosa_samples = iris[iris.iloc[:, -1] != 'Iris-setosa']
num_non_setosa_samples = non_setosa_samples.shape[0]
print(f"Number of samples with class different from 'Iris-setosa': {num_non_setosa_samples}")
```
