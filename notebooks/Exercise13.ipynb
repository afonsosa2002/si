### Exercise 13: TanhActivation and SoftmaxActivation classes

### 13.1) Add a new activation class named TanhActivation to the "activation" module.

```python
class TanhActivation:
    def forward(self, inputs):

        # Aplica a função tangente hiperbólica (tanh) aos inputs.
        self.inputs = inputs
        self.output = np.tanh(inputs)
        return self.output

    def backward(self, dvalues):
        
        # Calcula o gradiente da função tanh para retropropagação.
        self.dinputs = dvalues * (1 - self.output ** 2)
        return self.dinputs
```

### 13.2) Add a new activation class named SoftmaxActivation to the "activation" module.

```python
class SoftmaxActivation:
    def forward(self, inputs):
        
        # Calcula a ativação softmax normalizando os valores exponenciais dos inputs.
        self.inputs = inputs
        exp_values = np.exp(inputs - np.max(inputs, axis=1, keepdims=True))
        self.output = exp_values / np.sum(exp_values, axis=1, keepdims=True)
        return self.output

    def backward(self, dvalues):
        
        # Calcula o gradiente da função softmax para retropropagação.
        self.dinputs = np.empty_like(dvalues)
        
        for index, (single_output, single_dvalues) in enumerate(zip(self.output, dvalues)):
            single_output = single_output.reshape(-1, 1)
            jacobian_matrix = np.diagflat(single_output) - np.dot(single_output, single_output.T)
            self.dinputs[index] = np.dot(jacobian_matrix, single_dvalues)

        return self.dinputs
```
