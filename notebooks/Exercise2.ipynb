### 2.1) Remove all samples containing at least one null value 

```python
import numpy as np
    
    def dropna(self):
        
        # Verifica linhas de `X` sem valores NaN
        mask = ~np.isnan(self.X).any(axis=1)
        
        # Mantém apenas as linhas que não possuem NaN em `X`
        self.X = self.X[mask]
        self.y = self.y[mask]
        
        return self 
```

### 2.2) Replace all null values with another value or the mean or median of the feature/variable

```python
def fillna(self, value):
    
        # Calcula a média de cada coluna ignorando NaN
        if value == "mean":
            
            means = np.nanmean(self.X, axis=0)
            nan_mask = np.isnan(self.X)
            self.X[nan_mask] = np.take(means, np.where(nan_mask)[1])
            
        # Calcula a mediana de cada coluna ignorando NaN    
        elif value == "median":
            
            medians = np.nanmedian(self.X, axis=0)
            nan_mask = np.isnan(self.X)
            self.X[nan_mask] = np.take(medians, np.where(nan_mask)[1])
            
        # Substitui todos os NaN por um valor fixo fornecid   
        else:
            
            self.X[np.isnan(self.X)] = value
        
        return self  
```

### 2.3) Remove a sample by its index

```python
def remove_by_index(self, index):
    self.X = np.delete(self.X, index, axis=0)
    self.y = np.delete(self.y, index, axis=0)
    return self  
```
