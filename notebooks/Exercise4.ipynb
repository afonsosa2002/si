### Exercise 4: Implement the Cosine distance in a module 

```python
import numpy as np

def cosine_distance(x, y):
    x = np.array(x)
    y = np.array(y)

    # Verifica se as dimensões são iguais
    if x.ndim != 1 or y.ndim != 2 or x.shape[0] != y.shape[1]:
        return np.full(y.shape[0], np.nan)  # Retorna NaN se as dimensões não coincidem

    # Calcula o produto entre cada vetor 
    dot_product = np.dot(y, x)

    # Calcula a norma dos vetores
    x_norm = np.linalg.norm(x)
    y_norms = np.linalg.norm(y, axis=1)

    # Retorna NaN se qualquer vetor tiver 0
    if x_norm == 0 or np.any(y_norms == 0):
        return np.full(y.shape[0], np.nan)

    # Calcula a similaridade cosseno
    cosine_similarity = dot_product / (x_norm * y_norms)

    # A distância cosseno é 1 menos a similaridade cosseno
    return 1 - cosine_similarity
```
