### Exercise 6: Implementing stratified splitting

### 6.1) add the "stratified_train_test_split" function 

```python
import numpy as np
from typing import Tuple

def stratified_train_test_split(data, test_size: float, random_state: int = None) -> Tuple:
    
    # Define a semente para o gerador de números aleatórios
    if random_state is not None:
        np.random.seed(random_state)

    # Obtém os rótulos das classes 
    target = data.labels  

    # Encontra as classes únicas e a contagem de exemplos de cada classe
    unique_classes, class_counts = np.unique(target, return_counts=True)
    
    # Listas para armazenar os índices dos dados de treino e teste
    train_indices = []
    test_indices = []

    # Loop sobre cada classe para fazer a divisão estratificada
    for class_label, class_count in zip(unique_classes, class_counts):
        
        # Encontra os índices dos exemplos pertencentes à classe atual
        class_indices = np.where(target == class_label)[0]

        # Define o número de amostras de teste a partir da classe
        n_test_samples = int(class_count * test_size)

        # Escolhe aleatoriamente índices dessa classe
        np.random.shuffle(class_indices)

        # Separa os índices para o conjunto de teste e treino
        test_indices_class = class_indices[:n_test_samples]
        test_indices.extend(test_indices_class)
        train_indices_class = class_indices[n_test_samples:]
        train_indices.extend(train_indices_class)

    # Seleciona os dados de treino e teste 
    train_data = data.select(train_indices)  
    test_data = data.select(test_indices)  

    # Retorna os dados de treino e teste
    return train_data, test_data
```

### 6.2) Test the "stratified_train_test_split" function with the iris dataset.

```python
import numpy as np
from sklearn.datasets import load_iris
from typing import Tuple

# Classe para armazenar os dados 
class Dataset:
    def __init__(self, data, labels):
        self.data = data
        self.labels = labels

    def select(self, indices):
        return Dataset(self.data[indices], self.labels[indices])
    
    
def stratified_train_test_split(data: Dataset, test_size: float, random_state: int = None) -> Tuple[Dataset, Dataset]:
    if random_state is not None:
        np.random.seed(random_state)

    target = data.labels  
    unique_classes, class_counts = np.unique(target, return_counts=True)
    
    train_indices = []
    test_indices = []

    for class_label, class_count in zip(unique_classes, class_counts):
        class_indices = np.where(target == class_label)[0]
        n_test_samples = int(class_count * test_size)
        np.random.shuffle(class_indices)

        test_indices_class = class_indices[:n_test_samples]
        test_indices.extend(test_indices_class)

        train_indices_class = class_indices[n_test_samples:]
        train_indices.extend(train_indices_class)

    train_data = data.select(train_indices)
    test_data = data.select(test_indices)

    return train_data, test_data

# Carrega o conjunto de dados Iris
iris = load_iris()
data = iris.data
labels = iris.target

# Cria o objeto Dataset
dataset = Dataset(data, labels)

# Divide os dados 
train_data, test_data = stratified_train_test_split(dataset, test_size=0.2, random_state=42)

# Exibe os tamanhos dos conjuntos de dados
print(f"Train data shape: {train_data.data.shape}")
print(f"Test data shape: {test_data.data.shape}")
```
